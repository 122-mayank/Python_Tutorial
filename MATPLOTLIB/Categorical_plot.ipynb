Categorical Data Plots


boxplot , violinplot , stripplot , swarmplot, barplot , countplot

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('tips')
df
```

```python
df.rename(columns={"sex":"gender"},inplace=True)
df
```

```python
sns.countplot(x = df['gender'] , hue =df['smoker'])
```

```python
sns.barplot(x = df['gender'], y = df['total_bill'])
```

```python
import numpy as np
sns.barplot(x = df['gender'], y = df['total_bill'] , estimator=np.sum)
```

```python
sns.boxplot(x ='tip',y='day',data = df)
```
