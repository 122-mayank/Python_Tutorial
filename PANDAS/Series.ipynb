Pandas Series

A series is one dimensional labeled array capable of holding any data type.The axis labels are collectively called the index.

```python
import numpy as np
import pandas as pd
```

```python
labels = ['a' , 'b' , 'c']
my_list = [10 , 20 , 30]
my_array = np.array([10 , 20 , 30])
d = {'a':10 , 'b':20 , 'c':30}
```

```python
pd.Series(my_list)
```

```python
pd.Series(my_list , index = labels)
```

```python
pd.Series(my_array)
```

```python
pd.Series(d)
```
