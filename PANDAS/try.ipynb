```python

#The syntax for a 2D NumPy array is:
#arr[row_slice, column_slice]
arr = np.array([
    [10 , 20 , 30],
    [40 , 50 , 60],
    [70 , 80 , 90]
    ])
print(arr[0:2,1:])
```

```python
"""
For a 2D array:

axis=0 → work down the rows → gives column-wise result
axis=1 → work across the columns → gives row-wise result
"""
arr = np.array([
    [10 , 20 , 30],
    [40 , 50 , 60],
    [70 , 80 , 90]
    ])
print(arr.sum(axis = 1))
```

```python
arr = np.array([
    [10 , 20 , 30],
    [40 , 50 , 60],
    [70 , 80 , 90]
    ])
print(arr.sum(axis = 0))
```

```python
arr = np.array([
    [10 , 20 , 30],
    [40 , 50 , 60],
    [70 , 80 , 90]
    ])
print(arr.max(axis = 1))
```

```python
arr = np.array([
    [10 , 20 , 30],
    [40 , 50 , 60],
    [70 , 80 , 90]
    ])
print(arr.mean(axis = 0))
```
