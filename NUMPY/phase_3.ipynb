```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
## Data Set :[restaurant_id , 2021 , 2022 , 2023 ,2024]
sales_data = np.array([
    [1 , 150000 , 180000 , 220000 , 250000],
    [2 , 120000 , 140000 , 160000 , 190000],
    [3 , 200000 , 230000 , 260000 , 300000],
    [4 , 180000 , 210000 , 240000 , 270000],
    [5 , 160000 , 185000 , 205000 , 230000]
])

print("=== Zomato slales analysis ===")
print("\n Shape of the data " , sales_data.shape)
print("\n Sample data for 3 first restau: ",sales_data[: , 1:])
```

```python
## total sales per year
print(np.sum(sales_data , axis = 0))
yearly_total = np.sum(sales_data[: , 1:] , axis = 0)
print("Yearly total ", yearly_total)

## Minimum saled per restaurant
min_sales = np.min(sales_data[: , 1:] , axis = 1)
print("Min Sales", min_sales)

## Maximum sales per year
max_sales = np.max(sales_data[: , 1:] , axis = 0)
print("Max Sales " , max_sales)

## Avg sales per restaurant
avg_sales = np.mean(sales_data[: , 1:] , axis = 1)
print("Avg sales" , avg_sales)
```

```python
cumsum = np.cumsum(sales_data[: , 1:] , axis = 1)
print("Cumulative sum" , cumsum)

plt.figure(figsize=(6 , 6))
plt.plot(np.mean(cumsum , axis = 0))
plt.title("Average cumulative sales per year across restaurant")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
vector1 = np.array([1 ,2 , 3 , 4 , 5 ])
vector2 = np.array([6 , 7 , 8, 9 , 10])

print("Additon of arrays " , vector1 + vector2)
print("\n Multiplication of vectors " , vector1 * vector2)
print("\n Dot Product " , np.dot(vector1 , vector2))

angle = np.arccos(np.dot(vector1 , vector2) / (np.linalg.norm(vector1) * np.linalg.norm(vector2)) )
print("Angle ", angle)
```
