```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
array1 = np.array([[1 , 2, 3] , [4, 5 , 6]])
array2 = np.random.rand(3 , 3)
array3 = np.zeros((4 ,4))
```

```python
np.save('array1.npy' , array1)
np.save('array2.npy' , array2)
np.save('array3.npy' , array3)
```

```python
loaded_array1 = np.load('array1.npy')
print(loaded_array1)
```

```python
logo = plt.imread('numpy_logo.png')

# Step 2: Save the image data as a NumPy .npy file
np.save('numpy_logo.npy', logo)

print("PNG image converted to NumPy .npy file successfully!")

# Step 3: Load the .npy file
logo_loaded = np.load('numpy_logo.npy')

# Step 4: Display the image
plt.figure(figsize=(10, 5))

plt.subplot(121)
plt.imshow(logo_loaded)
plt.title("NumPy Logo")
plt.grid(False)

plt.subplot(122)
plt.imshow(logo)
plt.title("Original PNG")
plt.grid(False)

plt.show()
```
